Final Project – Student Record Management System

```python

# Program 46: Student Record Management System Using CSV

import csv
import os

# CSV file name
filename = "students.csv"

# Function to add student
def add_student():
    sid = input("Enter Student ID: ")
    name = input("Enter Student Name: ")
    age = input("Enter Student Age: ")
    course = input("Enter Course: ")

    students = []

    # Read existing records
    if os.path.exists(filename):
        with open(filename, "r", newline="") as file:
            students = list(csv.DictReader(file))

    # Check duplicate ID
    if any(s["ID"] == sid for s in students):
        print("Student ID already exists!")
        return

    # Add new student
    with open(filename, "a", newline="") as file:
        writer = csv.writer(file)

        if not os.path.exists(filename) or os.path.getsize(filename) == 0:
            writer.writerow(["ID", "Name", "Age", "Course"])

        writer.writerow([sid, name, age, course])

    print("Student added successfully!")


# Function to view students
def view_students():
    if not os.path.exists(filename):
        print("No student records found!")
        return

    with open(filename, "r", newline="") as file:
        reader = csv.DictReader(file)

        print("\n--- Student Records ---")

        records = list(reader)

        if not records:
            print("No records available!")
            return

        for student in records:
            print(student)


# Function to search student
def search_student():
    sid = input("Enter Student ID to search: ")

    if not os.path.exists(filename):
        print("No records found!")
        return

    with open(filename, "r", newline="") as file:
        reader = csv.DictReader(file)

        for student in reader:
            if student["ID"] == sid:
                print("Student Found:", student)
                return

    print("Student not found!")


# Function to update student
def update_student():
    sid = input("Enter Student ID to update: ")

    if not os.path.exists(filename):
        print("No records found!")
        return

    with open(filename, "r", newline="") as file:
        students = list(csv.DictReader(file))

    found = False

    for student in students:
        if student["ID"] == sid:
            student["Name"] = input("Enter New Name: ")
            student["Age"] = input("Enter New Age: ")
            student["Course"] = input("Enter New Course: ")
            found = True
            break

    if found:
        with open(filename, "w", newline="") as file:
            writer = csv.DictWriter(
                file, fieldnames=["ID", "Name", "Age", "Course"]
            )
            writer.writeheader()
            writer.writerows(students)

        print("Student updated successfully!")
    else:
        print("Student not found!")


# Function to delete student
def delete_student():
    sid = input("Enter Student ID to delete: ")

    if not os.path.exists(filename):
        print("No records found!")
        return

    with open(filename, "r", newline="") as file:
        students = list(csv.DictReader(file))

    updated_students = [
        student for student in students if student["ID"] != sid
    ]

    if len(updated_students) == len(students):
        print("Student not found!")
        return

    with open(filename, "w", newline="") as file:
        writer = csv.DictWriter(
            file, fieldnames=["ID", "Name", "Age", "Course"]
        )
        writer.writeheader()
        writer.writerows(updated_students)

    print("Student deleted successfully!")


# Main menu
while True:
    print("\n===== STUDENT RECORD MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:
            add_student()
        elif choice == 2:
            view_students()
        elif choice == 3:
            search_student()
        elif choice == 4:
            update_student()
        elif choice == 5:
            delete_student()
        elif choice == 6:
            print("Thank you for using the system!")
            break
        else:
            print("Invalid choice! Try again.")

    except ValueError:
        print("Please enter a valid number.")
```
